# I. Data descriptive

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
file_path = r'D:\Data\GridProd_1922_monthly.dta'
data = pd.read_csv(file_path)
```

```python
filtered_data = data.drop_duplicates(subset=['name_prod', 'Year', 'Month'])
filtered_data = filtered_data.dropna(subset=["Iron_Prod", "Steel_Prod"])
```

```python
X = filtered_data["Steel_Prod"].values.reshape(-1, 1)  
y = filtered_data["Iron_Prod"].values  

model = LinearRegression()
model.fit(X, y)
```

## I.1 The relationship between pig iron and crude steel

```python

x_range = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
y_pred = model.predict(x_range)

plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='blue', alpha=0.6, label='Data points')
plt.plot(x_range, y_pred, color='red', label='Fitted line')
plt.title('Scatter Plot with Fitted Line')
plt.xlabel('Crude Steel Production (in tpa)')
plt.ylabel('Iron Production (in tpa)')
plt.legend(loc='upper left', bbox_to_anchor=(1, 1)) 
plt.grid(True)
plt.show()
```

```python
slope = model.coef_[0]
intercept = model.intercept_

plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='blue', alpha=0.6, label='Data points')
plt.plot(x_range, y_pred, color='red', label='Fitted line')
plt.title('Scatter Plot with Fitted Line and Regression Equation')
plt.xlabel('Crude Steel Production (in tpa)')
plt.ylabel('Iron Production (in tpa)')
plt.legend(loc='upper left', bbox_to_anchor=(1, 1)) 
plt.grid(True)

equation_text = f"y = {slope:.2f}x + {intercept:.2f}"
plt.text(X.min(), y.max(), equation_text, fontsize=12, color='red', bbox=dict(facecolor='white', alpha=0.5))
plt.savefig('D:/Figures/Correlation_Production_2019_2022.png', format='png', dpi=300, bbox_inches='tight')
plt.show()
```

## I.2 Area of plant from global steel tracker

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import rcParams
file_path = r"D:\Data\PlantGridInfo.xlsx"
data = pd.read_excel(file_path)
```

```python
rcParams['font.family'] = 'Times New Roman'
unique_data = data.drop_duplicates(subset='PlantID')

all_areas = unique_data['Area']

operating_plants = unique_data[unique_data['Capacityop'].str.contains('operating', case=False, na=False)]
operating_areas = operating_plants['Area']

all_median = all_areas.median()
all_mean = all_areas.mean()
operating_median = operating_areas.median()
operating_mean = operating_areas.mean()

plt.figure(figsize=(12, 6))

plt.hist(all_areas, bins=30, alpha=0.5, label='All Plants', color='blue')
plt.axvline(all_median, color='blue', linestyle='--', label=f"All Median: {all_median:.2f}")
plt.axvline(all_mean, color='blue', linestyle='-', label=f"All Mean: {all_mean:.2f}")
plt.hist(operating_areas, bins=30, alpha=0.7, label='Operating Plants', color='orange')
plt.axvline(operating_median, color='orange', linestyle='--', label=f"Operating Median: {operating_median:.2f}")
plt.axvline(operating_mean, color='orange', linestyle='-', label=f"Operating Mean: {operating_mean:.2f}")
plt.title("Steel Plant Area Distribution: All vs Operating", fontsize=14)
plt.xlabel("Area (sq. km)", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.legend()
plt.grid(alpha=0.3)

output_path = r'D:\Figures\Steel_Plant_Area_Distribution.pdf'
plt.savefig(output_path, format='pdf', dpi=800)

plt.show()
```
